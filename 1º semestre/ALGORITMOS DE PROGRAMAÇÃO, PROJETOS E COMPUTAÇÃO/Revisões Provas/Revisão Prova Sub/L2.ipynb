Elabore um programa que leia e armazene 10 números reais, o programa deverá informar

• A média dos elementos maiores que 30

• A quantidade de números pares e maiores que 24 (incluindo 24).

• Imprimir a lista lida

```python
lista = []
soma_maior30 = 0
qtd_maior30 = 0
qtd_par_maior24 = 0

for i in range (10):
    num = float(input(f"Digite o {i+1}º número: "))
    lista.append(num)

    if num % 2 == 0 and num >= 24:  
        qtd_par_maior24 += 1
    
    if num > 30:
        soma_maior30 += num
        qtd_maior30 += 1

if qtd_maior30 > 0:
    media = soma_maior30 / qtd_maior30
else:
    media = 0

print(f"A média dos números maiores que 30 é: {media:.2f}")
print(f"Quantidade de números pares maiores que 24: {qtd_par_maior24}")
print(f"A lista é {lista}")
```

Elabore um programa que leia um número inteiro e uma função que calcule o seu fatorial

```python
def calcular_fatorial(n):
    fatorial = 1
    for i in range(1, n + 1):
        fatorial *= i
    return fatorial 

numero = int(input("Digite um número inteiro para calcular o fatorial: "))

if numero < 0:
    print("O fatorial não é definido para números negativos.")
else:
    resultado = calcular_fatorial(numero)
    print(f"O fatorial de {numero} é: {resultado}")
```

10. Faça um programa com uma função chamada somaImposto. A função possui dois parâmetros
formais: taxaImposto, que é a quantia de imposto sobre vendas expressa em porcentagem, e custo,
que é o custo de um item antes do imposto. A função “altera” o valor de custo para incluir o imposto
sobre vendas.

```python
def somaImposto (taxaImposto, custo):
    novo_valor = (custo * (taxaImposto/100)) + custo
    return novo_valor

tax = float(input("Digite a taxa: "))
custo = float(input("Digite o custo: "))

valor_novo = somaImposto(tax, custo)
print(f"O valor com imposto é: R$ {valor_novo:.2f}")
```

Calculadora de IMC:

• Crie uma função chamada calc1 que receba o peso (em kg) e a altura (em metros) de uma
pessoa. A função deve calcular o Índice de Massa Corporal (IMC) e retornar o resultado.

• IMC = peso / (altura * altura)

• Chame a função passando os argumentos por posição

```python
def calc1(peso, altura):
    resultado = peso / (altura * altura)
    return resultado

peso = float(input("Digite seu peso (em kg): "))
altura = float(input("Digite sua altura (em metros): "))

imc = calc1(peso, altura)
print(f"Seu IMC é {imc:.2f}")
```

12. Desenvolva uma função gera3 que receba um valor inicial, um valor final e um passo (com valor
padrão de 1). A função deve retornar uma lista contendo a sequência de números gerada

```python
def gera3 (valor_inicial, valor_final, passo=1):
    lista = []
    for i in range (valor_inicial, valor_final+1, passo):
        lista.append(i)
    return lista

vi = int(input("Digite o valor inicial: "))
vf = int(input("Digite o valor final: "))

calc = gera3(vi, vf)
print(calc)
```

14. Elabore um programa que leia e armazene numa lista 20 preços de celulares. O programa deverá

```python
soma_cel = 0
maior_valor = 0
qtd_mais3k = 0
lista = []

for i in range (20):
    cel = float(input(f"Digite o preço do {i+1}º celular: R$ "))
    lista.append(cel)

for i in lista:
    soma_cel += i

    if i > maior_valor:
        maior_valor = i
    
    if i > 3000:
        qtd_mais3k += 1
    

print(f"A média dos preços de celular é: R${soma_cel / 20:.2f}")
print(f"O maior valor de celular foi: R${maior_valor:.2f}")
print(f"Há {qtd_mais3k} celulares com preço superior a R$ 3000")
```

15. Foram anotadas as idades e alturas de 30 alunos. Faça um Programa que determine quantos
alunos com mais de 13 anos possuem altura inferior à média de altura desses alunos.

```python
qtd_alunos = 0
soma_alturas = 0
lista_altura = []
lista_idade = []

for i in range (30):
    idade = int(input(f"Digite a {i+1}ª idade: "))
    altura = float(input(f"Digite a {i+1}ª altura (metros): "))

    lista_altura.append(altura)
    lista_idade.append(idade)

for altura in lista_altura:
    soma_alturas += altura

media = soma_alturas / 30

for i in range(30):
    if lista_idade[i] > 13 and lista_altura[i] < media:
        qtd_alunos += 1

print(qtd_alunos)
```

16. Construa um programa que leia dois números inteiros: a e b e uma lista com N valores inteiros. O
programa deverá imprimir quantos elementos pertencem ao intervalo [a;b]

```python
qtd_lista = int(input("Digite quantos números você quer inserir na lista: "))
a = int(input("Digite o valor de A (inicial): "))
b = int(input("Digite o valor de B (final): "))
lista = []

for i in range(qtd_lista):
    n = int(input(f"Digite o {i+1}º valor: "))
    lista.append(n)

cont = 0

for valor in lista:
    if a <= valor <= b:
        cont += 1

print(f"Quantidade de elementos no intervalo [{a};{b}]: {cont}")
```

20. Escreva uma função, chamada contaVoigais, que conta todas as vogais presentes no texto
recebido como parâmetro e retorna um dicionário contendo a quantidade de cada vogal. Escreva
um programa a fim de testar sua função, e exiba, no fim, os dados do dicionário retornado.

```python
def contaVogais(texto):
    cont = {'a': 0, 'e': 0, 'i': 0, 'o': 0, 'u': 0}
    for i in texto.lower():
        if i in cont:
            cont[i] += 1
    return cont

texto = input("Digite o seu texto: ")
resultado = contaVogais(texto)

print("Quantidade de vogais encontradas:")
print(resultado)
```

22. Construa um programa que seja constituído de uma lista GAB de 10 elementos caracteres ( esta
lista pode ser constituída somente dos caracteres a, b, c, d e e. O programa irá ler o nome e a
resposta de 10 alunos de uma turma e deverá imprimir a nota de cada aluno (considerando que
cada questão vale 1,0 ponto). O programa deverá também imprimir a média da sala


```python
print("Digite o gabarito das 10 questões (letras de 'a' a 'e'):")
gabarito = []
for i in range(10):
    while True:
        resposta = input(f"Questão {i+1}: ").lower()
        if resposta in ['a', 'b', 'c', 'd', 'e']:
            gabarito.append(resposta)
            break
        else:
            print("Resposta inválida. Use apenas 'a', 'b', 'c', 'd' ou 'e'.")

notas = []

for i in range(10):
    print(f"\nAluno {i+1}:")
    nome = input("Nome: ")
    respostas_aluno = []

    for j in range(10):
        while True:
            resposta = input(f"Resposta da questão {j+1}: ").lower()
            if resposta in ['a', 'b', 'c', 'd', 'e']:
                respostas_aluno.append(resposta)
                break
            else:
                print("Resposta inválida. Use apenas 'a', 'b', 'c', 'd' ou 'e'.")

    nota = 0
    for r_aluno, r_gabarito in zip(respostas_aluno, gabarito):
        if r_aluno == r_gabarito:
            nota += 1

    notas.append(nota)
    print(f"Nota de {nome}: {nota:.1f}")

media = sum(notas) / len(notas)
print(f"\nMédia da turma: {media:.2f}")
```
