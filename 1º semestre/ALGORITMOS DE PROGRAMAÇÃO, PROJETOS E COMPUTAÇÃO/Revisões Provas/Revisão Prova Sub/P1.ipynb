Faça um programa que receba o preço de venda e a quantidade vendida de 100 produtos. Calcule e mostre:

A quantidade de produtos que tiveram venda maior que 30 unidades

A média do preço de venda dos produtos com vendas entre 10 e 50 unidades

A porcentagem de produtos com preço de venda superior a R$ 100,00 em relação ao total

```python
venda_maior30 = 0
soma_preco_10_50 = 0
cont_10_50 = 0
venda_superior100 = 0

for i in range(100):
    preco = float(input(f"Preço do {i+1}º produto: "))
    qtd = int(input(f"Quantidade de vendas do {i+1}º produto: "))

    if qtd > 30:
        venda_maior30 += 1
    
    if 10 <= qtd <= 50:
        soma_preco_10_50 += preco
        cont_10_50 += 1
    
    if preco > 100:
        venda_superior100 += 1

if cont_10_50 > 0:
    media = soma_preco_10_50 / cont_10_50
else:
    media = 0

porcentagem = (venda_superior100 / 100) * 100

print(f"A quantidade de produtos que tiveram vendas maior que 30 unidades foi {venda_maior30}")
print(f"A média do preço de vendas dos produtos que tiveram entre 10 e 50 unidades foi {media:.2f}")
print(f"A porcentagem de produtos que tiveram preço superior a R$ 100 foi {porcentagem:.2f}%")
```

Escreva um programa que leia o custo e a quantidade disponível de 100 itens em estoque. Depois, calcule e exiba:

A quantidade de produtos com menos de 10 itens em estoque

A média do custo dos produtos que possuem entre 15 e 40 unidades em estoque

A porcentagem de produtos cujo custo seja maior que R$ 50,00

```python
menor10 = 0
custo15_40 = 0
soma15_40 = 0
custo_maior50 = 0

for i in range (100):
    custo = float(input(f"Digite o preço de custo do {i+1}º produto: "))
    qtd = int(input(f"Digite a quantidade em estoque do {i+1}º produto: "))

    if qtd < 10:
        menor10 += 1
    
    if 15 <= qtd <= 40:
        custo15_40 += custo
        soma15_40 += 1
    
    if custo > 50:
        custo_maior50 += 1

media = custo15_40 / soma15_40

porcentagem = (custo_maior50 / 100) * 100

print(f"A quantidade de produtos com menos de 10 unidades em estoque é de {menor10} produtos.")
print(f"A média de custo dos produtos com 15 a 40 unidades em estoque é de R${media:.2f}.")
print(f"A porcentagem de produtos com custo maior que R$50,00 é de {porcentagem:.2f}%")
```

Faça um programa que receba o preço de custo e o número de unidades vendidas de 100 produtos. Calcule e mostre:

Quantos produtos tiveram vendas superiores a 70 unidades

A média do preço de custo dos produtos com vendas entre 30 e 60 unidades

A porcentagem de produtos com preço de custo inferior a R$ 20,00 em relação ao total

```python
venda_superior70 = 0
custo30_60 = 0
soma30_60 = 0
custo_inferior20 = 0

for i in range (100):
    custo = float(input(f"Digite o preço de custo do {i+1}º produto: R$ "))
    vendas = int(input(f"Digite a quantidade de vendas do {i+1}º produto: "))

    if vendas > 70: 
        venda_superior70 += 1
    
    if 30 <= vendas <= 60:
        custo30_60 += custo
        soma30_60 += 1
    
    if custo < 20: 
        custo_inferior20 += 1

if soma30_60 > 0:
    media = custo30_60 / soma30_60
else:
    media = 0
    
porcentagem = (custo_inferior20 / 100) * 100

print(f"A quantidade de vendas superior a 70 unidades é de {venda_superior70} produtos.")
print(f"A média de custo dos produtos com 30 a 60 vendas é dada por R${media:.2f}.")
print(f"A porcentagem de produtos com custo inferior a R$20 é de {porcentagem:.2f}.")
```
