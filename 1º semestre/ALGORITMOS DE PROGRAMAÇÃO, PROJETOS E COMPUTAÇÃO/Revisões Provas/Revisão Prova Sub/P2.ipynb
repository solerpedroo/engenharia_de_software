elabore um programa que leia 10 numeros inteiros e insira numa lista. o programa deverá possuir uma função que tenha como argumento a lista e retorna o maior valor par dessa lista.

```python
def maior_par(lista):
    maior = -9999999
    for i in lista:
        if (i > maior) and (i % 2 == 0):
            maior = i
    return maior

lista = []

for i in range (10):
    num = float(input(f"Digite o {i+1}º número: "))
    lista.append(num)

maior = maior_par(lista)
print(maior)
```

Elabore um programa que leia 10 números inteiros e insira em uma lista. O programa deverá possuir uma função que receba a lista como argumento e retorne a quantidade de números negativos que ela possui.

```python
def num_negativo(lista):
    cont = 0
    for nums in lista:
        if nums < 0:
            cont += 1
    return cont

lista = []
for i in range (10):
    num = float(input(f"Digite o {i+1}º número da lista: "))
    lista.append(num)

qtd_negativos = num_negativo(lista)
print(f"Quantidade de números negativos: {qtd_negativos}")
```

Elabore um programa que leia 8 nomes de cidades e insira em uma lista. O programa deverá possuir uma função que receba a lista como argumento e retorne o nome da cidade com o maior número de caracteres.

```python
def cidade(lista):
    maior = ''
    for i in lista:
        if len(i) > len(maior):
            maior = i
    return maior


lista = []
for i in range (8):
    cidade = input("Digite o nome da cidade: ")
    lista.append(cidade)

qtd_cidade = cidade(lista)
print(f"A cidade com maior número de carácteres é : {qtd_cidade}")
```

Elabore um programa que leia 5 preços de produtos e insira em uma lista. O programa deverá possuir uma função que receba a lista como argumento e retorne o preço mais baixo.

```python
def menor(lista):
    menor = 0
    for i in lista:
        if i < menor:
            menor = i
    return menor

lista = []

for i in range (5):
    preco = float(input(f"Digite o preço do {i+1}º produto: R$"))
    lista.append(preco)

menor_valor = menor(lista)
print(f"O menor valor é: R${menor_valor:.2f}")
```

Elabore um programa que leia 6 idades de pessoas e insira em uma lista. O programa deverá possuir uma função que receba a lista como argumento e retorne a média das idades.

```python
def idades (lista):
    soma_idades = 0
    for i in lista:
        soma_idades += i
    return soma_idades

lista = []

for i in range (6):
    idade = int(input(f"Digite a idade da {i+1}ª pessoa: "))
    lista.append(idade)

media = idades(lista) / len(lista)

print(f"A média da idade das pessoas é igual a {media:.0f}")
```

```python
def letra_a (lista):
    cont = 0
    for i in range(len(lista)):
        if lista[i][0].lower() == 'a':
            cont += 1
    return cont

lista = []

for i in range (7):
    alunos = input(f"Digite o nome do {i+1}º aluno: ")
    lista.append(alunos)

letra = letra_a(lista)
print(f"A quantidade de alunos que começam com a letra 'A' é {letra}")
```

```python
def nomes (lista):
    nome = lista[0]
    for names in lista:
        if len(names) < len(nome): 
            nome = names
    return nome

lista = []

for i in range (8):
    nome = input(f"Digite o {i+1}º nome: ")
    lista.append(nome)

menos_caracteres = nomes(lista)
print(f"O nome com menos caracteres é: {menos_caracteres}")
```
