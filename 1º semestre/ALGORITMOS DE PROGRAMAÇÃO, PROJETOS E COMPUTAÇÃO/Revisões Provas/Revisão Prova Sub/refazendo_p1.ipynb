elabore um programa que leia um numero x e imprima a soma de todos os elementos múltiplos de 9 entre e 1 e x(inclusive), se for necessário use o comando for.
ex: se x=30 a soma dos multiplos de 9 será 54

```python
soma = 0
x = int(input("Digite um valor inteiro: "))

for i in range(1, x+1, 1):
    if i % 9 == 0:
        soma += i

print(soma)
```

elabore um programa que imprima a quantidade de números entre 535 e 1350 e quando divididos por 7 retornem como resto um número menor ou igual a 4, se necessário use o comando for

```python
qtd = 0

for i in range (535, 1351):
    if i % 7 <= 4:
        qtd += 1

print(qtd)
```

faça um programa que receba o preço de custo e a quantidade de itens em estoque de 100 produtos. calcule e mostre:

a quantidade de produtos com mais de 20 itens em estoque

a média de preço de custo dos produtos que possuem entre 50 e 80 produtos em estoque 

a porcentagem de produtos com preço de estoque inferior a R$ 35,50 todos os produtos analisados

```python
qtd_20mais = 0
soma50_80 = 0
qtd50_80 = 0
menor_35 = 0

for i in range(100):
    custo = float(input(f"Digite o preço de custo do {i+1}º produto: "))
    qtd_estoque = int(input(f"Digite a quantidade de itens em estoque do {i+1}º produto: "))

    if qtd_estoque > 20:
        qtd_20mais += 1
    
    if qtd_estoque >= 50 and qtd_estoque <= 80: 
        soma50_80 += custo
        qtd50_80 += 1
    
    if custo < 35.5:
        menor_35 += 1

if qtd50_80 > 0:
    media = soma50_80 / qtd50_80
else:
    media = 0

porcentagem = (menor_35/100)*100

print(qtd_20mais)
print(media)
print(porcentagem)
```
