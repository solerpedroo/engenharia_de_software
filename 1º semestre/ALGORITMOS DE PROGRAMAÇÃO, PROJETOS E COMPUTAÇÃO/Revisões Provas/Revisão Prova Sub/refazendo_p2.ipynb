elabore um programa que leia 10 numeros inteiros e insira numa lista. o programa deverá possuir uma função que tenha como argumento a lista e retorna o maior valor par dessa lista.


```python
def maior_par(lista):
    maior = -999999
    for i in lista:
        if (i % 2 == 0) and (i > maior):
            maior = i
    return maior

lista = []

for i in range(10):
    num = int(input(f"Digite o {i+1}º número: "))
    lista.append(num)

resultado = maior_par(lista)

print(resultado)
```

Elabore um programa que leia um dicionário de livros, onde o código do livro é a chave do dicionário. Além do código o livro possui com informações o título, nome do autor e ano de lançamento. Construa um programa que leia 10 livros e insira no dicionário. O programa não deve permiritr a inserção de códigos repetidos


```python
dic = {}
cont = 0

while cont < 10: 
    codigo = int(input("Digite o código do livro: "))
    if codigo not in dic.keys():
        lista = []
        titulo = input("Digite o título do livro: ")
        lista.append(titulo)
        nome = input("Digite o nome do autor: ")
        lista.append(nome)
        ano = int(input("Digite o ano de lançamento do livro: "))
        lista.append(ano)
        dic[codigo] = lista
        cont += 1
    else:
        print("Código já existente")
```

Cálculo de média da PUC com arredondamento

```python
def arredondar_nota_puc(nota):
    """
    Arredonda uma nota de acordo com as regras da PUC (inteiros ou meios).
    Exemplos:
    4.75 a 5.24 = 5.0
    5.25 a 5.74 = 5.5
    5.75 a 6.24 = 6.0
    """
    parte_inteira = int(nota)
    parte_decimal = nota - parte_inteira

    if parte_decimal < 0.25:
        return float(parte_inteira)
    elif parte_decimal < 0.75:
        return parte_inteira + 0.5
    else:
        return float(parte_inteira + 1)

def calcular_media_final():
    print("--- Calculadora de Média Final da PUC ---")

    # Solicitar as notas ao usuário
    try:
        teste1 = float(input("Digite a nota do Teste 1: "))
        print(f"Nota do teste 1: {teste1}")
        teste2 = float(input("Digite a nota do Teste 2: "))
        print(f"Nota do teste 2: {teste2}")
        prova1 = float(input("Digite a nota da Prova 1: "))
        print(f"Nota da prova 1: {prova1}")
        prova2 = float(input(f"Digite a nota da Prova 2: "))
        print(f"Nota da prova 2: {prova2}")
        pi = float(input("Digite a nota do P.I.: "))
        print(f"Nota do P.I: {pi}")
    except ValueError:
        print("Entrada inválida. Por favor, digite apenas números para as notas.")
        return

    # Calcular a média dos testes
    media_teste = (teste1 + teste2) / 2

    # Calcular a média final sem arredondamento inicial
    media_final_bruta = (prova1 * 0.35) + (prova2 * 0.4) + (media_teste * 0.15) + (pi * 0.10)

    # Arredondar a média final usando a função específica da PUC
    media_final_arredondada = arredondar_nota_puc(media_final_bruta)

    print(f"\n--- Resultados ---")
    print(f"Média dos Testes: {media_teste:.2f}")
    print(f"Média Final (sem arredondamento): {media_final_bruta:.2f}")
    print(f"Média Final (arredondada PUC): {media_final_arredondada:.1f}")

# Executar o programa
if __name__ == "__main__":
    calcular_media_final()
```
