```python
#Exercício 1

salario = float(input("Digite o seu salário atual: "))

plano = input("Digite o seu plano de trabalho (1, 2, 3): ")

aumentos = {
    "1": 0.10,
    "2": 0.15,
    "3": 0.20
}

if plano in aumentos:
    novo_salario = salario + (salario * aumentos[plano])
    print(f"O novo salário é de R$ {novo_salario:.2f}")
else:
    print(f"Plano não existente, digite um novo número de 1 a 3.")
```

```python
#Exercício 2

x = int(input("Digite o valor da base: "))  
n = int(input("Digite o valor do expoente: "))  

resultado = 1

for i in range(n):
    resultado *= x

print(f"O resultado de {x}^{n} é {resultado}")
```

```python
#Exercício 4

n = int(input("Quantos números deseja inserir? "))

soma_pares = 0
contagem_impares = 0

for i in range(n):
    numero = int(input(f"Digite o {i+1}º número: "))

    if i % 2 == 0:
        soma_pares += numero
    else:
        contagem_impares += 1

print(f"\nSoma dos números pares: {soma_pares}")
print(f"\nA quantidade de números ímpares digitados: {contagem_impares}")
```

```python
#Exercício 5

while True:
    a = float(input("Digite o valor de a: "))
    b = float(input("Digite o valor de b: "))
    c = float(input("Digite o valor de c: "))
    d = float(input("Digite o valor de d: "))
    e = float(input("Digite o valor de e: "))
    f = float(input("Digite o valor de f: "))

    if a == 0 and b == 0 and c == 0 and d == 0:
        print("Encerrando o programa")
        break
    
    denominador = (a * e - b * d)

    if denominador == 0:
        print(f"O sistema não tem solução (divisão por zero)")
        break
    else:
        x = (c * e - b * f) / denominador
        y = (a * f - c * d) / denominador
        print(f"A Solução do sistema: x = {x:.2f}, y = {y:.2f}")
        break
```

```python
#Exercício 1

nota1 = float(input("Digite a nota 1 do aluno: "))
nota2 = float(input("Digite a nota 2 do aluno: "))

peso1 = int(input("Digite o peso 1: "))
peso2 = int(input("Digite o peso 2: "))

media_ponderada = ( (nota1 * peso1) + (nota2 * peso2) ) / (peso1 + peso2)

if media_ponderada >= 7 and media_ponderada < 10:
    print(f"Aluno Aprovado! Com a média {media_ponderada:.1f}")
if media_ponderada < 7:
    print(f"Aluno Reprovado! Com a média {media_ponderada:.1f}")
if media_ponderada == 10:
    print(f"Aluno Aprovado com Distinção! Com a média {media_ponderada:.1f}") 
```

```python
#Exercício 2

n_int_1 = int(input("Digite o primeiro número inteiro: "))
n_int_2 = int(input("Digite o segundo número inteiro: "))
n_int_3 = int(input("Digite o terceiro número inteiro: "))

if n_int_1 > n_int_2 and n_int_1 > n_int_3:
    print(f"O número {n_int_1} é o maior")
if n_int_2 > n_int_1 and n_int_2 > n_int_3:
    print(f"O número {n_int_2} é o maior")
if n_int_3 > n_int_2 and n_int_3 > n_int_1:
    print(f"O número {n_int_3} é o maior")
```

```python
#Exercício 3

N = int(input("Digite um número inteiro: "))

resultado = 1

if N == 0:
    resultado = 1
else:
    for i in range(1, N + 1):
        resultado *= i

print(f"O fatorial de {N} é {resultado}")
```

```python
#Exercício 4

N = int(input("Digite o número de elementos (N) a serem digitados (deve ser maior ou igual a 10): "))

if N < 4:
    print("O valor de N deve ser maior ou igual a 10.")
else:
    maior = None
    menor = None

    for i in range(N):
        x = int(input(f"Digite o número {i+1}: "))

        if maior is None or x > maior:
            maior = x
        if menor is None or x < menor:
            menor = x

    print(f"O maior número digitado foi: {maior}")
    print(f"O menor número digitado foi: {menor}")
```
